## Question 3

Minimize $f = x^2 + y^2 + z^2$ s.t. $x + 2y + z = 6$.

**1. Lagrangian:** $\mathcal{L} = x^2 + y^2 + z^2 - \lambda(x + 2y + z - 6)$

**2. System:**
$$2x - \lambda = 0,\quad 2y - 2\lambda = 0,\quad 2z - \lambda = 0,\quad x+2y+z=6$$
$$\Rightarrow x = \tfrac{\lambda}{2},\; y = \lambda,\; z = \tfrac{\lambda}{2}$$

**3. Solve:** $\tfrac{\lambda}{2} + 2\lambda + \tfrac{\lambda}{2} = 6 \Rightarrow \lambda = 2 \Rightarrow (x,y,z) = (1,2,1)$, $f = \mathbf{6}$

**4. Classify:** Minimum ($f$ = squared distance from origin, bounded below).

**5. Constraint role:** Restricts solution to the plane; forces it off the origin (unconstrained min = 0) to the nearest plane point, giving 6. $\lambda = 2$ = shadow price.

## Question 5


Minimize $\|y - Xw\|^2 + \lambda\|w\|^2$.

$$J(w) = (y-Xw)^\top(y-Xw) + \lambda w^\top w$$
$$\nabla_w J = -2X^\top y + 2X^\top X w + 2\lambda w = 0$$
$$\boxed{\hat{w} = (X^\top X + \lambda I)^{-1} X^\top y}$$

**Change vs. OLS** ($\hat{w} = (X^\top X)^{-1}X^\top y$): adds $\lambda I$ before inverting.
**Impact:** makes $X^\top X + \lambda I$ always invertible (stability with collinear features); shrinks weights toward 0 (reduces overfitting). $\lambda=0$ → OLS; larger $\lambda$ → more shrinkage.

```python
import torch

rs = 0.00887
r = torch.tensor(30000.1, requires_grad=True)

# τ = 2r · sqrt(r/rs) · sqrt(1 - rs/r)
tau = 2 * r * torch.sqrt(r / rs) * torch.sqrt(1 - rs / r)
tau.backward()

print("tau  =", tau.item())
print("dtau/dr =", r.grad.item())
```

```python
!wget -O rav4.csv https://raw.githubusercontent.com/rahulbhadani/Datasets/master/RAV4/2020-04-15-05-12-00_2T3Y1RFV8KC014025_CAN_Messages_5_Hz.csv

import pandas as pd
df = pd.read_csv("rav4.csv")
print("Shape:", df.shape)
print("Columns:", list(df.columns))
df.head()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

steer_col = "steer_angle"   # <-- fix to match real column
yaw_col   = "yaw_rate"      # <-- fix to match real column

data = df[[steer_col, yaw_col]].dropna()
X = data[[steer_col]].values
y = data[yaw_col].values

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

model = LinearRegression()
model.fit(X_train, y_train)

print("Coefficient (slope w1):", model.coef_[0])
print("Intercept (w0):", model.intercept_)
print("R^2 on test data:", r2_score(y_test, model.predict(X_test)))
```
